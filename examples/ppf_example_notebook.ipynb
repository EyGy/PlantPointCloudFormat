### Read and view PPF files

```python
#read a ppf file from path
verbose_example = ppf.read_ppf(filepath="full_ppf_example_begonia.ply")
```

```python
#display the ppf metadata
verbose_example.metadata
```

### Writing PPF files

```python
#Modifying the PPF metadata
verbose_example.metadata["plant_category"] = "dicot"
verbose_example.metadata["growth_stage"] = "vegetative"

#Example of writing PPF file to ascii PLY
ppf.write_ppf(verbose_example,"verbose_example_ascii.ply", encoding="ascii")
```
