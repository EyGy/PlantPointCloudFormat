### Writing PPF files

````python
#Modifying the PPF metadata
ppf_example_plant.metadata["plant_category"] = "dicot"
ppf_example_plant.metadata["growth_stage"] = "vegetative"

#Example of writing PPF file to ascii PLY
ppf.write_ppf(ppf_example_plant,"ppf_example_ascii.ply", encoding="ascii")
````

### Opening, editing and re-saving point clouds in tools like CloudCompare might:
   - Discard all PPF comment headers
   - Rename properties (e.g., semantic_id → scalar_semantic_id)
   - Convert int properties to float/double
   - Apply coordinate shifts

To reverse this always keept the original ppf file and use this function to restore ppf format:
    ```ppf.transfer_metadata(edited, original, output)```

````python
ppf.transfer_metadata("cloudcompare_edited_example.ply","original_example_begonia_ppf.ply","example_begonia_ppf.ply",re_center=True)
````
